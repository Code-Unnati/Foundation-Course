### Lab: Handling Missing and Inconsistent Data using Pandas in Python

#### Problem Statement:
Real-world datasets often contain missing or inconsistent data, which can impact data analysis and machine learning models. This lab focuses on identifying, handling, and resolving missing and inconsistent data using Pandas.


#### Objective:

- Handle missing values using different strategies (removal, imputation).
- Detect and resolve inconsistent data.

**Part 1: Importing Libraries and Loading Data**


```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# Load the dataset
df = pd.read_csv("Dataset/sample_data.csv")
print("Dataset Loaded Successfully")
df.head()
```

**Part 2: Identifying Missing Values**

```python
# Display missing values count per column
print("Missing values per column:")
print(df.isnull().sum())
```

**Part 3: Handling Missing Values**


```python
# Removing rows with missing values
print("Removing rows with missing values:")
df_dropped = df.dropna()
print(df_dropped.head())
```

```python
# Filling missing values with a specific value
print("Filling missing values with zero:")
df_filled_zero = df.fillna(0)
print(df_filled_zero.head())
```

```python
# Filling missing values with column mean
print("Filling missing values with column mean:")
df_filled_mean = df.fillna(df.mean(numeric_only=True))
print(df_filled_mean.head())
```

**Part 4: Detecting and Resolving Inconsistent Data**


```python
# Checking unique values in a categorical column
df['Category'] = df['Category'].str.strip()  # Remove leading/trailing spaces
print("Unique values in 'Category' column before fixing:")
print(df['Category'].unique())
```

```python
# Standardizing text values
df['Category'] = df['Category'].replace({'Electr0nics': 'Electronics', 'elec': 'Electronics'})
print("Unique values in 'Category' column after fixing:")
print(df['Category'].unique())
```
