### Lab Title: Statistical Analysis of Gym Equipment Usage




#### Problem Statement:
A gym tracks the average number of times a particular treadmill is used per hour. Historically, the treadmill is used 3 times per hour on average. The gym wants to analyze:
1. What is the probability that the treadmill will be used exactly 5 times in an hour?
2. What is the probability that it will be used more than 6 times in an hour?
3. Simulate treadmill usage over 24 hours and visualize the distribution.



#### Objective:
- Use statistical techniques to analyze treadmill usage.
- Compute probabilities using the Poisson distribution.
- Simulate treadmill usage over a 24-hour period and visualize the results.



#### Dataset:
The dataset is taken from 'https://www.kaggle.com/datasets/bipulnath98/treadmill-users-dataset/data' 



#### Requirements:
- Python installed (version 3.x recommended).
- Pandas, NumPy, SciPy, and Matplotlib libraries installed (`pip install pandas numpy scipy matplotlib`).
- Jupyter Notebook (optional but recommended for running the lab).



#### Implementation:


```python

# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python

# Load dataset
df = pd.read_csv("treadmil-users.csv")
```

```python

# Define parameters
average_usage_per_hour = 3  # Given mean usage rate
```

```python

# 1. Probability that the treadmill is used exactly 5 times in an hour
prob_5_uses = poisson.pmf(5, average_usage_per_hour)

# 2. Probability that the treadmill is used more than 6 times in an hour
prob_more_than_6 = 1 - poisson.cdf(6, average_usage_per_hour)

# 3. Simulating treadmill usage over 24 hours
np.random.seed(42)
simulated_usage = np.random.poisson(average_usage_per_hour, size=24)
```

```python
# Print results
print(f"Probability that the treadmill is used exactly 5 times in an hour: {prob_5_uses:.4f}")
print(f"Probability that the treadmill is used more than 6 times in an hour: {prob_more_than_6:.4f}")
```

```python

# Visualization: Histogram of treadmill usage simulation
plt.figure(figsize=(8, 5))
plt.hist(simulated_usage, bins=range(0, max(simulated_usage) + 2), edgecolor="black", alpha=0.7)
plt.xlabel("Number of Uses per Hour")
plt.ylabel("Frequency")
plt.title("Simulated Treadmill Usage Over 24 Hours")
plt.xticks(range(0, max(simulated_usage) + 1))
plt.show()
```
