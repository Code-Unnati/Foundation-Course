### Lab Title: Real-Time Hypothesis t-tests on Fitness Center Data



#### Problem Statement:
A fitness center wants to analyze whether there is a significant difference in the average calories burned during workouts between two groups of individuals. Hypothesis testing using a t-test will be used to determine if the difference is statistically significant.



#### Objective:
- Understand hypothesis testing and t-tests.
- Compare two groups to determine if there is a significant difference in their average calories burned.
- Apply statistical analysis using Python.



#### Dataset:
The dataset is taken from [Kaggle](https://www.kaggle.com/datasets/salihobaid/fitbit-datasets-0312201605122016)




#### Requirements:
- Python installed (version 3.x recommended).
- Pandas, NumPy, and SciPy libraries installed (`pip install pandas numpy scipy`).
- Jupyter Notebook (optional but recommended for running the lab).



#### Implementation:


```python

# Import necessary libraries
import pandas as pd
import numpy as np
from scipy import stats
```

```python

# Load dataset
df = pd.read_csv("dailyActivity.csv")
```

```python

# Create two groups based on activity level
median_steps = df['TotalSteps'].median()
df['Group'] = np.where(df['TotalSteps'] >= median_steps, 'High Activity', 'Low Activity')
```

```python

# Separate the two groups
group_high = df[df['Group'] == 'High Activity']['Calories']
group_low = df[df['Group'] == 'Low Activity']['Calories']
```

```python

# Perform independent t-test
t_stat, p_value = stats.ttest_ind(group_high, group_low, equal_var=False)
```

```python

# Print results
print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant difference in calories burned between the two groups.")
else:
    print("Fail to reject the null hypothesis: No significant difference in calories burned between the two groups.")
```
