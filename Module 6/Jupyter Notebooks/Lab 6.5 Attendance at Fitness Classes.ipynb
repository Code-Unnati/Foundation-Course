### Lab Title: Statistical Analysis of Member Attendance at Fitness Classes



#### Problem Statement:
A gym tracks member attendance for a group fitness class. Historically, 60% of members who sign up for the class actually attend. The gym wants to analyze:
1. What is the probability that exactly 7 members will show up?
2. What is the probability that fewer than 5 members will show up?
3. Simulate attendance for 30 classes and visualize the distribution.



#### Objective:
- Use statistical techniques to determine attendance probabilities.
- Compute probabilities using the binomial distribution.
- Simulate class attendance and visualize the results.



#### Dataset:
The dataset is taken from `fitness_class_attendance` 



#### Requirements:
- Python installed (version 3.x recommended).
- Pandas, NumPy, SciPy, and Matplotlib libraries installed (`pip install pandas numpy scipy matplotlib`).
- Jupyter Notebook (optional but recommended for running the lab).



#### Implementation:


```python

# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python

# Load dataset
df = pd.read_csv("fitness_class_attendance.csv")  
```

```python

# Define parameters
total_members = 10  # Members signed up per class
attendance_prob = 0.6  # Given probability of attendance
```

```python

# 1. Probability that exactly 7 members attend
prob_7_attend = binom.pmf(7, total_members, attendance_prob)

# 2. Probability that fewer than 5 members attend
prob_less_than_5 = binom.cdf(4, total_members, attendance_prob)

# 3. Simulating attendance for 30 classes
np.random.seed(42)
simulated_attendance = np.random.binomial(total_members, attendance_prob, size=30)
```

```python

# Print results
print(f"Probability that exactly 7 members will show up: {prob_7_attend:.4f}")
print(f"Probability that fewer than 5 members will show up: {prob_less_than_5:.4f}")

```

```python

# Visualization: Histogram of attendance simulation
plt.figure(figsize=(8, 5))
plt.hist(simulated_attendance, bins=np.arange(3, 11) - 0.5, edgecolor="black", alpha=0.7)
plt.xlabel("Number of Attendees")
plt.ylabel("Frequency")
plt.title("Simulated Attendance for 30 Classes")
plt.xticks(range(3, 11))
plt.show()
```
