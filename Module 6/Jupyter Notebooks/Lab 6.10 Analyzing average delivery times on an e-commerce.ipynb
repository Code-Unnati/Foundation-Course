### Lab Title: Real-Time Hypothesis z-tests on E-Commerce Platform Delivery Times



#### Problem Statement:
An e-commerce platform claims that the average delivery time for orders is 48 hours. The objective of this study is to analyze real-time order data and conduct a z-test to determine whether the actual average delivery time significantly differs from the claimed 48 hours.



#### Objective:
- Understand hypothesis testing using a z-test.
- Verify the company's claim regarding average delivery time.
- Apply statistical analysis using Python.



#### Dataset:
The dataset is taken from [Kaggle](https://www.kaggle.com/datasets/baneetsinghsachdeva/ecommerce)




#### Requirements:
- Python installed (version 3.x recommended).
- Pandas, NumPy, and statsmodels libraries installed (`pip install pandas numpy statsmodels`).
- Jupyter Notebook (optional but recommended for running the lab).



#### Implementation:


```python

# Import necessary libraries
import pandas as pd
import numpy as np
from statsmodels.stats.weightstats import ztest
```

```python

# Load dataset
df = pd.read_csv("ecommerce_dataset.csv")
```

```python

# Generate synthetic delivery times for demonstration purposes
df['delivery_time'] = np.random.normal(loc=50, scale=10, size=len(df))  
# Replace with actual delivery time column if available
```

```python

# Perform one-sample z-test
claimed_mean = 48
z_stat, p_value = ztest(df['delivery_time'], value=claimed_mean)
```

```python

# Print results
print("Z-Statistic:", z_stat)
print("P-Value:", p_value)
```

```python

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The actual average delivery time significantly differs from 48 hours.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in the average delivery time.")
```
