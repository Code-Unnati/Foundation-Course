#### Steps


#### Step 1: Define the Graph Class
We use a dictionary to represent the graph with adjacency lists.

```python
# Step 1: Define the Graph class using an adjacency list
class Graph:
    def __init__(self):
        self.graph = {}  # Dictionary to store adjacency list

    def add_edge(self, u, v):
        # Adds an edge from u to v (for directed graph)
        if u in self.graph:
            self.graph[u].append(v)
        else:
            self.graph[u] = [v]
```

#### Step 2: Implement DFS Traversal Function
Use recursion and a set to track visited nodes.

```python
    def dfs_util(self, node, visited):
        visited.add(node)
        print(node, end=' ')
        for neighbor in self.graph.get(node, []):
            if neighbor not in visited:
                self.dfs_util(neighbor, visited)

    def dfs(self, start_node):
        visited = set()
        self.dfs_util(start_node, visited)
```

#### Step 3: Create a Graph and Add Edges

```python
# Create the graph and add edges
g = Graph()
g.add_edge(0, 1)
g.add_edge(0, 2)
g.add_edge(1, 2)
g.add_edge(2, 0)
g.add_edge(2, 3)
g.add_edge(3, 3)
```

#### Step 4: Perform DFS Traversal

```python
# Perform DFS starting from node 2
g.dfs(2)
```
