### Objective
To implement a function in Python that swaps two numbers without using a third variable, using the bitwise XOR operator.

#### Problem Statement
Write a Python function to swap two numbers using the XOR (^) operator without using any third or temporary variable.

#### Steps

#### Step 1: Initialize Two Variables

```python
# Initial values
a = 10
b = 20
print(f"Before swap: a = {a}, b = {b}")
```

#### Step 2: Swap Using XOR Operator

```python
# Swapping using XOR
a = a ^ b
b = a ^ b
a = a ^ b

print(f"After swap: a = {a}, b = {b}")
```

#### Step 3: Wrap It in a Function

```python
def xor_swap(x, y):
    print(f"Before swap: x = {x}, y = {y}")
    x = x ^ y
    y = x ^ y
    x = x ^ y
    print(f"After swap: x = {x}, y = {y}")
    return x, y
```

```python
# Test
xor_swap(5, 7)
```

#### Conclusion

In this lab, we demonstrated how to use the XOR bitwise operator to swap two integers without using a third variable.
XOR is reversible. Each bit in the binary representation of the numbers is flipped in a way that allows us to retrieve the original values in reverse order.
This technique is memory efficient and useful in environments where using an extra variable is not preferred.
