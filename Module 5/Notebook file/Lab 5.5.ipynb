### Finding GCD Using Euclidean Algorithm

#### Objective

To implement the Euclidean algorithm in Python to find the Greatest Common Divisor (GCD) of two integers.

#### Equipment Required

- Computer with Python 3.x installed
- Jupyter Notebook or any Python IDE

#### Prerequisites

- Basic Python programming knowledge
- Understanding of loops and conditionals
- Basic number theory concept: GCD

#### Problem Statement

Write a Python function to compute the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

#### Steps

#### Step 1: Understand the Euclidean Algorithm

The Euclidean algorithm finds the GCD of two numbers based on the principle:

GCD(a, b) = GCD(b, a % b)

Where % is the modulo operator. The algorithm repeats this process until b becomes 0, and then a is the GCD.

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a
```

```python
# Function execution
num1 = 56
num2 = 98
print(f"GCD of {num1} and {num2} is {gcd(num1, num2)}")
```

#### Step 3: Explanation

- The function takes two integers, a and b.
- It uses a while loop that runs until b becomes zero.
- In each iteration, we replace a with b, and b with a % b.
- When b becomes zero, a holds the GCD.
- The function then returns a.

```python
print(f"GCD of 48 and 18 is {gcd(48, 18)}")
print(f"GCD of 101 and 103 is {gcd(101, 103)}") 
print(f"GCD of 270 and 192 is {gcd(270, 192)}")
```

#### Conclusion

The Euclidean algorithm efficiently computes the Greatest Common Divisor (GCD) of two numbers using repeated modulo operations. It is one of the oldest and fastest methods for finding the GCD and forms the basis of many number theory applications.
