###  Longest Repeating Substring using Dynamic Programming


#### Objective

#### Problem Statement
Given a string `s`, find the longest repeating substring using dynamic programming.

A repeating substring is a substring that appears more than once in the string without overlapping.

#### Steps


##### Step 1: Define the Input String

```python
# Sample input string
s = "Programming"
```

#### Step 2: Initialize the DP Table
We use a 2D table where `dp[i][j]` stores the length of the longest common suffix of substrings `s[0…i-1]` and `s[0…j-1]`.
Ensure that `i != j` to avoid overlapping substrings.

```python
def longest_repeating_substring(s):
    n = len(s)
    dp = [[0] * (n + 1) for _ in range(n + 1)]
    max_len = 0
    end_index = 0

    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if s[i - 1] == s[j - 1] and i != j:
                dp[i][j] = dp[i - 1][j - 1] + 1
                if dp[i][j] > max_len:
                    max_len = dp[i][j]
                    end_index = i

    return s[end_index - max_len:end_index]
    
```

##### Step 3: Call the Function and Display the Result


```python
result = longest_repeating_substring(s)
print(f" Longest repeating substring: '{result}'")
```
